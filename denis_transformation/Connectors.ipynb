```python
# Function to connect ADLS with Databricks using Account Key method
def adls_connect():
    spark.conf.set(
    "fs.azure.account.key.adlsdatadenis.dfs.core.windows.net",
    dbutils.secrets.get(scope="denis", key="AdlsAccessKey"))
    return "ADLS Gen2 Connected Successfully"
```

```python
# Function to list all the files in Bronze layer in ADLS Gen2
def list_bronze_files():
    display(dbutils.fs.ls("abfss://denis@adlsdatadenis.dfs.core.windows.net/medallion/bronze/"))
    return "Bronze files listed successfully"
```
