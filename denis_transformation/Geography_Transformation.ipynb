### Transformation

```python
# Removing column '_id' as it is not required
geo_df = geo_df.drop("_id")

```

```python
display(geo_df.limit(5))
```

```python
check_string_value_as_nan(geo_df)
```

```python
# Adding a Surrogate key to uniquely identify each rows
#geo_df = geo_df.withColumn("GeoKey", monotonically_increasing_id())
window_spec = Window.orderBy(F.lit(1))

# 2. Assign the sequential surrogate key
geo_df = geo_df.withColumn("GeoKey", F.row_number().over(window_spec))
```

```python
display(geo_df)
```

```python
geo_df = geo_df.select("GeoKey","Country","Town")
```
