### Transformation

```python
# Splitting Location as Country and Town
sales_df = sales_df \
    .withColumn("Country", F.split(F.col("Location"), ";").getItem(0)) \
    .withColumn("Town", F.split(F.col("Location"), ";").getItem(1))
```

```python
display(sales_df.limit(5))
```

```python
# Date Formatting
sales_df = sales_df.withColumn("Date", 
    F.date_format(F.to_date(F.col("Date"), "yyyy-MM-dd"), "dd-MM-yyyy"))
```

```python
display(sales_df.limit(5))
```

```python
# Deleting a Column
sales_df = sales_df.drop("Location")
```

```python
display(sales_df.limit(5))
```

```python
# Selected Columns
sales_df = sales_df.select("fSalesPrimaryKey", "ProductID", "SalesRepID", "Date","Country", "Town", "Units", "PercentOfStandardCost", "RevenueDiscount")
```
