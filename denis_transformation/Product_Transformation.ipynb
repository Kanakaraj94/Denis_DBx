### Transformation

```python
# Removing duplicate Product IDs
product_df = product_df.groupBy("ProductID","ProductName","Sub Category Key","Color","RetailPrice","StandardCost").agg(
    F.count("ProductID").alias("DistinctProductID")
)
```

```python
# Selected columns after removing duplicate IDs
product_df = product_df.select("ProductID","ProductName","Sub Category Key","Color","RetailPrice","StandardCost")
```

```python
product_df = product_df.sort(F.col("ProductID"), asc=True)
```
